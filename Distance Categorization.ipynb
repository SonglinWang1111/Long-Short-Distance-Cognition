## Distance Categorization

This experiment asks 4 human raters and employs 3 LLMs across multiple configurations to classify the type of human migration discussed in each paper into long-distance human migration or short-distance human migration, or both/neither.

It's important to note that this task differs from more straightforward NER, but relatively comprehensive. The conceptualization of migration type is rarely explicitly addressed by the context, but should be inferred according to more implicit natural language descriptions.

### General commands for getting response

We employ 3 different models: gpt-4.1-mini, gemini-2.5-flash, and deepseek-3.2-chat via API provided by them. For each model, we consider 3 scenarios: 0-shot, 1-shot, and 3-shot, which enables us to see if giving examples can make any difference (if they can learn something from a person).

```python
def api_def(provider):
    # API keys are hidden in another file
    with open('api.txt', 'r', encoding='utf-8') as file:
        lines = [line.strip() for line in file.readlines()]

        # Different LLMs need different kinds of command
        if provider == 'chatgpt':
            client = OpenAI(api_key=str(lines[0]), base_url="https://api.openai.com/v1")
            return client
        
        elif provider == 'deepseek':
            client = OpenAI(api_key=str(lines[1]), base_url="https://api.deepseek.com")
            return client
        
        elif provider == 'gemini':
            genai.configure(api_key=str(lines[4]))
```

```python
def get_completion(prompt, provider, model, temperature):
    messages = [{"role": "system", "content": "You are a human geographer, please simply classify the text according to requirements."},
               {"role": "user", "content": prompt}]

    # Different LLMs need different kinds of command
    if provider == 'gpt':
        client = api_def('chatgpt')
        response = client.responses.create(
            model=model,
            input=messages,
            temperature=temperature,
            max_output_tokens = MAX_TOKEN
        )
        return response.output_text
        
    elif provider == 'ds':
        client = api_def('deepseek')
        response = client.chat.completions.create(
            model=model,
            messages=messages,
            temperature=temperature,
            max_tokens = MAX_TOKEN
        )
        return response.choices[0].message.content

    elif provider == 'gemini':
        api_def('gemini')
        model_gemini = genai.GenerativeModel(model)
        response = model_gemini.generate_content(
            prompt,
            generation_config={"temperature": temperature, "max_output_tokens": MAX_TOKEN}
        )
        return response.text        
```

```python
CONTEXT = f"""
Please classify the article based on the type of human migration it address by assigning one of the following codes: \n
'1': It focuses on or observes evidence of long-distance human migration. \n
'2': It focuses on or observes evidence of short-distance human migration. \n
'3': It focuses on or observes evidence of both long-distance and short-distance human migration. \n
'0': It does not mention anything about human migration distance. \n \n

Instructions for doing classification: \n
1.  While classifying an article, focus only on its own content. Do not infer any missing information from e.g., past \
tasks. Only concentrate on sections that explicitly describe this exact work, such as 'abstract', 'objectives', \
'data', 'methods', 'results', and 'conclusions', etc., rather than those that primarily summarize others' \
work, like 'introduction', 'literature review' or 'related works' etc. \n
2.  Classification of migration type should be based on interpreting context, rather than only extracting explicit mentions of \
'long-distance human migration' or 'short-distance human migration'. \n \n

Do not infer any missing information. Output only the single character without explanations, opinions or meta-comments.
"""
```

```python
def get_response(paper: dict, example, provider, model, temperature):

    fail_count = 0
    while True:
        if example == 3:
            prompt = f"""
            <Group leader>: {CONTEXT} \n
            Now please assign a character for the following article: \n
            \"\"\"{shot[0]}\"\"\"
            \n
            <Human geographer>: My answer is: 1 \n
            <Group leader>: Please assign a character for another article: \n
            \"\"\"{shot[1]}\"\"\"
            \n
            <Human geographer>: My answer is: 3 \n
            <Group leader>: Please assign a character for another article: \n
            \"\"\"{shot[2]}\"\"\"
            \n
            <Human geographer>: My answer is: 2 \n
            
            <Group leader>: Please assign a character for this article like what the human geographer did: \n"
            f"{paper}\n"
            "Keep in mind that output only one character."
            """

        elif example == 1:
            prompt = f"""
            <Group leader>: {CONTEXT} \n
            Now please assign a character for the following article: \n
            \"\"\"{shot[0]}\"\"\"
            \n
            <Human geographer>: My answer is: 1 \n
            
            <Group leader>: Please assign a character for this article like what the human geographer did: \n"
            f"{paper}\n"
            "Keep in mind that output only one character."
            """

        elif example == 0:
            prompt = f"""
            <Group leader>: {CONTEXT} \n
            Please assign a character for the following article: \n
            f"{paper}\n"
            "Keep in mind that output only one character."
            """

        response = get_completion(prompt, provider, model, temperature)
        return response
```

### Organization

For each model and each scenario, we consider 5 different temperatures and iterate each experiment for 10 times to reduce the influence of hallucinated outputs. The whole process is applied to 30 different papers in the same way.

```python
def response_agg(literature, example, provider, model, idx):
    results = []
    fail_count = 0
    temperatures = np.round(np.arange(0.0, 2.0, 0.4), 1)
        
    for t in tqdm(temperatures, desc=f"File {idx+1}"):        
        for run_id in range(10):
            success = False
            while not success:
                try:
                    result = {"answer": get_response(literature, example, provider, model, t)}
                    result["provider"] = provider
                    result["model"] = model
                    result["few shot"] = example
                    result["temperature"] = t
                    result["run"] = run_id
                    results.append(result)
                    # When exceed Tokens per minute
                    time.sleep(1)
                    success = True
                except (RateLimitError, APIError, APITimeoutError, ResourceExhausted) as e: # AuthenticationError, BadRequestError, NotFoundError
                    time.sleep(1)
                    fail_count = fail_count + 1
                    if fail_count >= 10:
                        raise ValueError(f"Can't get response, retried too many times") from e

    df = pd.DataFrame(results)
    return df
```

```python
def batch_response_to_df(all_literature, example, provider, model):

    save_path = f"results_e1/{provider}_{example}.csv"
    os.makedirs("results_e1", exist_ok=True)

    if os.path.exists(save_path):
        temporal_df = pd.read_csv(save_path)
        processed_ids = set(temporal_df["file_id"].astype(int).tolist())
        print(f"Start from paper {len(processed_ids)+1}")
    else:
        temporal_df = pd.DataFrame()
        processed_ids = set()
        
    all_results = [temporal_df] if not temporal_df.empty else []
    
    for idx, literature in enumerate(tqdm(all_literature.values(), desc="Progress")):
        file_id = idx + 1
        if file_id in processed_ids:
            continue
        
        df = response_agg(literature, example, provider, model, idx)
        df["file_id"] = file_id
        all_results.append(df)

        temporal_df = pd.concat(all_results, ignore_index=True)
        temporal_df.astype(str).to_csv(save_path, index=False)

    final_df = pd.concat(all_results, ignore_index=True)
    return final_df
```

In all: 30 documents, 3 models, 3 scenarios regarding amount of examples, 5 temperatures, 10 iterations.

### Analysis

We simply calculate the agreement.

#### Inter human rater

```python
manual_1 = pd.read_excel('manual_1.xlsx')
manual_1["class"] = (manual_1["Long distance"].astype(int) + manual_1["Short distance"].astype(int)*2).astype(str)
manual_1 = manual_1.drop(index=range(0, 3), columns=["ID", "AUTHOR", "TITLE", "Long distance", "Short distance"]).reset_index(drop=True)
```

```python
manual_2 = pd.read_excel('manual_2.xlsx')
manual_2["class"] = (manual_2["Long distance"].astype(int) + manual_2["Short distance"].astype(int)*2).astype(str)
manual_2 = manual_2.drop(index=range(0, 3), columns=["ID", "AUTHOR", "TITLE", "Long distance", "Short distance"]).reset_index(drop=True)
```

```python
manual_3 = pd.read_excel('manual_3.xlsx')
manual_3["class"] = (manual_3["Long distance"].astype(int) + manual_3["Short distance"].astype(int)*2).astype(str)
manual_3 = manual_3.drop(index=range(0, 3), columns=["ID", "AUTHOR", "TITLE", "Long distance", "Short distance"]).reset_index(drop=True)
```

```python
manual_4 = pd.read_excel('manual_4.xlsx')
manual_4["class"] = (manual_4["Long distance"].astype(int) + manual_4["Short distance"].astype(int)*2).astype(str)
manual_4 = manual_4.drop(index=range(0, 3), columns=["ID", "AUTHOR", "TITLE", "Long distance", "Short distance"]).reset_index(drop=True)
```

```python
manual_combined = pd.concat([manual_1, manual_2, manual_3, manual_4], axis=1, keys=['vec1', 'vec2', 'vec3', 'vec4'])

cols = manual_combined.columns
n = len(cols)
agreement_matrix = pd.DataFrame(np.eye(n), index=cols, columns=cols)

for col1, col2 in combinations(cols, 2):
    score = (manual_combined[col1] == manual_combined[col2]).mean()
    agreement_matrix.loc[col1, col2] = score
    agreement_matrix.loc[col2, col1] = score

plt.figure(figsize=(6, 5))
ax = sns.heatmap(agreement_matrix, annot=True, cmap='YlGnBu', vmin=0, vmax=1, fmt='.2f', square=True)
ax.set_title('Inter-rater agreement', fontsize=15)
ax.set_xlabel('')
ax.set_ylabel('')
labels_list = ['Participant 1', 'Participant 2', 'Participant 3', 'Participant 4']
ax.set_xticklabels(labels_list)
ax.set_yticklabels(labels_list, rotation=0)

filepath = os.path.join('plots/lit', 'ir_agree.png')
plt.savefig(filepath, dpi=300, bbox_inches='tight')

plt.show()
```

```python
from sklearn.metrics import cohen_kappa_score as cks
kappa_12 = cks(manual_1, manual_2)
kappa_13 = cks(manual_1, manual_3)
kappa_14 = cks(manual_1, manual_4)
kappa_23 = cks(manual_3, manual_2)
kappa_24 = cks(manual_4, manual_2)
kappa_34 = cks(manual_4, manual_3)

kappa_matrix = np.array([[1, kappa_12, kappa_13, kappa_14],
                        [kappa_12, 1, kappa_23, kappa_24],
                        [kappa_13, kappa_23, 1, kappa_34],
                        [kappa_14, kappa_24, kappa_34, 1]])

plt.figure(figsize=(6, 5))
ax = sns.heatmap(kappa_matrix, annot=True, cmap='YlGnBu', vmin=0, vmax=1, fmt='.2f', square=True)
ax.set_title('Cohen\'s kappa score', fontsize=15)
ax.set_xlabel('')
ax.set_ylabel('')
labels_list = ['Participant 1', 'Participant 2', 'Participant 3', 'Participant 4']
ax.set_xticklabels(labels_list)
ax.set_yticklabels(labels_list, rotation=0)

filepath = os.path.join('plots/lit', 'kappa.png')
plt.savefig(filepath, dpi=300, bbox_inches='tight')

plt.show()
```

#### Human-model

```python
df_dict = {}

for file_name in os.listdir('results_e1/'):
    if file_name.endswith('.csv'):
        file_path = os.path.join('results_e1/', file_name)

        df = pd.read_csv(file_path)
        df_name = file_name.replace('.csv', '')
        df_dict[df_name] = df

LLM_class = pd.concat(df_dict.values(), axis=0, ignore_index=True)
LLM_class['file_id'] = LLM_class['file_id'].astype(int) - 1
LLM_class['label'] = (LLM_class['file_id'].map(manual_1['class'])).astype(str).str.strip()
LLM_class['answer'] = LLM_class['answer'].astype(str).str.strip()
```

```python
LLM_class['accuracy'] = (LLM_class['label'] == LLM_class['answer']).astype(int)
LLM_new = LLM_class.drop(columns=['provider', 'answer', 'label']).reset_index(drop=True)
LLM_summary = LLM_new.groupby(['model', 'few shot', 'temperature']).agg({'accuracy': 'mean'}).reset_index()
LLM_summary
```

```python
# random
file_ids = np.repeat(np.arange(30), 10)
random_class = np.random.randint(0, 4, size=300)

random = pd.DataFrame({
    'file_id': file_ids,
    'answer': random_class
})
```

```python
random['label'] = (random['file_id'].map(manual_1['class'])).astype(str).str.strip()
random['answer'] = random['answer'].astype(str).str.strip()
random['accuracy'] = (random['label'] == random['answer']).astype(int)
random = random.drop(columns=['answer']).reset_index(drop=True)
random_summary = random['accuracy'].mean()
random_summary
```

```python
def plot(df, model, few_shot, pos):    
    subset = df[(df['model'] == model)]
    
    plt.figure(figsize=(8,6))
    plt.title(f"{model} - Agreement", fontsize=22, pad=15)

    few_shot_values = subset['few shot'].unique()
    colors = {'gpt-4.1-mini': '#036aff', 'deepseek-chat': '#9cafff', 'gemini-2.5-flash': '#ff7a40'}
    color = colors.get(model, '#036aff')
    line_styles = ['-', '--', '-.']

    plt.axhline(y=random_summary, color='black', linestyle=':', linewidth=2, label='random')
    for i, few_shot in enumerate(few_shot_values):
        few_shot_data = subset[subset['few shot'] == few_shot]
        line_style = line_styles[i % len(line_styles)]
        plt.plot(few_shot_data['temperature'], few_shot_data['accuracy'], marker='o', markersize=6, linewidth=2, \
                color=color, linestyle=line_style, label=f'{few_shot}-shot')
    
    plt.xlabel('Temperature', fontsize=14)
    plt.ylabel('Agreement', fontsize=14)
    plt.grid(True, alpha=0.3)
    plt.ylim(0.2, 0.6)
    plt.xticks(np.arange(0, 2, 0.4), fontsize=18)
    plt.yticks(np.arange(0.2, 0.7, 0.1), fontsize=18)
    plt.legend(title='Strategy', title_fontsize=14, fontsize=12, loc='best', framealpha=0.9)
    filepath = os.path.join(pos, f"{model}.png")
    plt.savefig(filepath, dpi=300, bbox_inches='tight')
    plt.close()
```

```python
for model in LLM_summary['model'].unique():
    for few_shot in LLM_summary['few shot'].unique():
        plot(LLM_summary, model, few_shot, 'plots/lit')
```

```python
def conf_matrix(df, model, few_shot, temperature):
    subset = df[(df['model'] == model) & (df['few shot'] == few_shot) & (df['temperature'] == temperature)]
    matrix = pd.DataFrame(sklearn.metrics.confusion_matrix(subset['label'], subset['answer']))
    return matrix
```

```python
for model in LLM_class['model'].unique():
    for few_shot in LLM_class['few shot'].unique():
        for temperature in LLM_class['temperature'].unique():
            conf_matrix(LLM_class, model, few_shot, temperature).to_csv(f"confusion_matrix/{model}_{few_shot}_{temperature}.csv", index=False)
```
